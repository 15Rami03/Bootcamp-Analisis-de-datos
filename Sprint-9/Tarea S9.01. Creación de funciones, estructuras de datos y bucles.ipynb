## 2. Conversor de temperaturas

Existen diversas unidades de temperatura utilizadas en distintos contextos y regiones. Las más comunes son Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) y Réaumur (°Re).

Selecciona al menos 2 conversores, de modo que al introducir una temperatura devuelva, como mínimo, dos conversiones, de modo que se puedan guardar (recuerda que un print() no se puede guardar nunca).

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado.

(EXTRA): Piensa una manera de almacenar todas las posibles conversiones en un solo objeto (¿Lista? ¿Diccionario? ¿DataFrame?) en vez de escribir muchos if else en función de la temperatura de origen y la temperatura de destino.

```python
#SOLUCION
unidades = {
    "C": {
        "nombre": "Celsius",
        "lim_inf": -273.15,
        "a_otras": lambda t: {
            "F": (t * 9/5) + 32,
            "K": t + 273.15
        }
    },
    "F": {
        "nombre": "Fahrenheit",
        "lim_inf": -459.67,
        "a_otras": lambda t: {
            "C": (t - 32) * 5/9,
            "K": ((t - 32) * 5/9) + 273.15
        }
    },
    "K": {
        "nombre": "Kelvin",
        "lim_inf": 0,
        "a_otras": lambda t: {
            "C": t - 273.15,
            "F": ((t - 273.15) * 9/5) + 32
        }
    }
}

unidades_permitidas = []
for clave in unidades:
    unidades_permitidas.append(clave)
    
def pedir_datos():
    print('Este es un conversor entre unidades de temperatura. \nIngresa la unidad de medida desde la que quieres hacer la conversion')

    while True:
        unidad = input(f'Selecciona entre las unidades: {unidades_permitidas}.').strip().upper()
        if unidad in unidades_permitidas:
            break
        print(f'Unidad no valida, solo se acepta: {unidades_permitidas}. Vuelve a intentarlo')

    while True:
        entrada = input(f'Ingresa la temperatura que quieres convertir').strip().replace(',','.')
        try:
            temperatura = float(entrada)
        except ValueError:
            print(f'Debes ingresar un número')
            break
        if temperatura < unidades[unidad]['lim_inf']:
            print(f'La temperatura ingresada para {unidad} no puede ser menor a: {unidades[unidad]['lim_inf']}')
            break

        return unidad, temperatura

def convertir(unidad, temperatura):
    conversiones = unidades[unidad]["a_otras"](temperatura )
    return conversiones

def conversor_temp():
    u, t = pedir_datos()
    resultado = convertir(u, t)
    
    print(f"\nTemperatura ingresada: {t}°{u}")
    print("Conversiones:")

    for destino, valor in resultado.items():
        print(f"→ {destino}: {round(valor, 2)}°{destino}")

conversor_temp()
```

## 3. Contador de palabras de un texto.

Escribe una función que dado un texto, muestre las veces que aparece cada palabra. Intenta que se gestionen todas las posibles casuísticas que hagan que el programa no funcione correctamente.

(EXTRA): ¿Cuál es la longitud media de las palabras del texto que has escrito? “¿Hola cómo va?” debería devolver (4+3+2) / 3 = 3

```python
def analizar_texto(texto):
    # 1) Pasar todo a minúsculas para evitar problemas con palabras repetidas
    texto = texto.lower()

    # 2) Cambiar cualquier símbolo raro por un espacio
    # (dejo letras y números; todo lo demás se transforma en espacio)
    filtro = ""
    for caracter in texto:
        if caracter.isalnum() or caracter == "á" or caracter == "é" or caracter == "í" or caracter == "ó" or caracter == "ú" or caracter == "ñ":
            filtro += caracter
        else:
            filtro += " "

    # 3) Separar en palabras (split elimina espacios dobles automáticamente)
    palabras = filtro.split()

    if not palabras:
        print("No se encontraron palabras en el texto.")
        return
    
    # 4) Contar cuántas veces aparece cada palabra
    contador = {}
    for palabra in palabras:
        if palabra not in contador:
            contador[palabra] = 1
        else:
            contador[palabra] += 1

    # 5) Mostrar las palabras y sus cantidades
    print("Conteo de palabras:")
    for palabra, veces in contador.items():
        print(f" - {palabra}: {veces} ")


    # EXTRA: longitud media de las palabras
    longitudes = [len(p) for p in palabras]
    media = sum(longitudes) / len(longitudes)
    print(f"\nLongitud media de las palabras: {round(media, 2)}")


# Ejemplo para probar:
analizar_texto("¡Hola!, hola, Hace mucho tiempo que no te he visto. Verdad?")

```

## 4. Diccionario inverso (con posibilidad de duplicados)

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa, es decir, intercambiando las claves y valores. Los valores y claves en el diccionario original son únicos; si éste no es el caso, la función debería imprimir un mensaje de advertencia, junto con una lista con los valores asociados a la clave repetida.

```python
def reverse_dictionary(dic):
    dic_inverso = {}

    for clave in dic:
        valor = dic[clave]

        # Si este valor no está, lo guardo con su clave
        if valor not in dic_inverso:
            dic_inverso[valor] = clave
        else:
            # Si el valor ya existía pero NO es lista, lo convierto
            if not isinstance(dic_inverso[valor], list):
                dic_inverso[valor] = [dic_inverso[valor]]

            # Añadir la nueva clave a la lista
            dic_inverso[valor].append(clave)

    return dic_inverso   

horario = {
    'Lunes': 'Matemática',
    'Martes': 'Catalán',
    'Miércoles': 'Castellano',
    'Jueves': 'Física',
    'Viernes': 'Economía'
    }

reverse_dictionary(horario)
```

##  1. Contador y ordenador de palabras de un texto.

El cliente quedó contento con el contador de palabras, pero ahora quiere leer archivos TXT y que calcule la frecuencia de cada palabra ordenadas dentro de las entradas habituales del diccionario según la letra con la que comienzan, es decir, las claves deben ir de la A a la Z y dentro de la A debemos ir de la A a Z. "tu_me_quieres_blanca.txt" la salida esperada sería:



```python
import string

def contar_palabras_por_letra(archivo):
    # Crear diccionario con claves 'a' a 'z'
    resultado = {letra: {} for letra in string.ascii_lowercase}

    # Leer el archivo
    with open(archivo, "r", encoding="utf-8") as f:
        texto = f.read().lower()

    # Reemplazar signos de puntuación por espacios
    limpiar = str.maketrans(string.punctuation + "¡¿", " " * (len(string.punctuation) + 2))
    texto = texto.translate(limpiar)

    # Separar en palabras
    palabras = texto.split()

    # Contar frecuencia dentro de cada letra inicial
    for palabra in palabras:
        inicial = palabra[0]

        if inicial in resultado:
            resultado[inicial][palabra] = resultado[inicial].get(palabra, 0) + 1

    # Ordenar palabras dentro de cada letra
    for letra in resultado:
        resultado[letra] = dict(sorted(resultado[letra].items()))

    return resultado


# --- EJEMPLO DE USO ---
archivo = r"C:\Users\ramir\OneDrive\Escritorio\Ramiro\Cursos IT Academy\Bootcamp-Analisis-de-datos\Sprint-9\Tú me quieres blanca, de Alfonsina Storni.txt"
frecuencias = contar_palabras_por_letra(archivo)

from pprint import pprint
pprint(frecuencias)
```

## 2. Conversión de tipos de datos.

El cliente recibe una lista de datos y necesita generar dos listas, 
la primera donde estarán todos los elementos que pudieron convertirse en flotantes y 
la otra donde están los elementos que no pudieron convertirse. 
Ejemplo de la lista que recibe el cliente:
['1.3', 'one', '1e10', 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
def conversion(datos):
    convertidos = []
    no_convertidos = []

    def intentar_convertir(valor):
        """Intenta convertir un único valor a float."""
        try:
            return float(valor)
        except:
            return None

    def procesar(elemento):
        """Procesa cualquier cosa: valor, lista o tupla."""
        if isinstance(elemento, (list, tuple)):
            for x in elemento:
                procesar(x)
        else:
            resultado = intentar_convertir(elemento)
            if resultado is None:
                no_convertidos.append(elemento)
            else:
                convertidos.append(resultado)

    # Procesamos todo lo que viene en la lista principal
    for item in datos:
        procesar(item)

    return convertidos, no_convertidos

datos= ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]
conversion(datos)
```

## 1. Generador de Contraseñas

Explora el funcionamiento del módulo random de la librería numpy. 🔗 Random module in NumPy

En este punto, el cliente ha detectado un problema con las contraseñas que utilizan sus trabajadores. Asdf1234, fechas de cumpleaños o similares. 
Para solucionarlo, nos ha encargado una función de Python que genere contraseñas más seguras. La función debe depender de los siguientes parámetros:

*longitud (int): Longitud de la contraseña
*mayúsculas (bool = True): Si tiene que aparecer mayúsculas
*minúsculas (bool = True): Si tiene que aparecer minúsculas
*números (bool = True): Si tiene que aparecer números
*signos (bool = False): Si tiene que aparecer caracteres especiales (,-$? o similares)
*Así pues, si ejecutamos la función de la siguiente forma:

crear_contraseña(10, True, True, True, True)

Deberíamos obtener un output (que debemos poder guardar) del estilo:

9Er,5Vn8P$

Asegúrate de que se cumplan todos los criterios, y que estas contraseñas sean realmente aleatorias.

*(EXTRA) 
Explora cómo podríamos hacer que la función copiara la contraseña automáticamente en el portapapeles del ordenador 
(como si la hubiéramos seleccionado y hecho ctrl+copy)

```python
import numpy as np
import pyperclip

def crear_contraseña(longitud, mayus=True, minus=True, numeros=True, signos=False, copiar=False):
    letras_mayus = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    letras_minus = "abcdefghijklmnopqrstuvwxyz"
    digitos = "0123456789"
    simbolos = "!@#$%&*?-_"
    
    bolsa = ""
    if mayus: bolsa += letras_mayus
    if minus: bolsa += letras_minus
    if numeros: bolsa += digitos
    if signos: bolsa += simbolos
    
    if not bolsa:
        print("No hay caracteres disponibles.")
        return None
    
    contraseña = "".join(np.random.choice(list(bolsa), size=longitud))
    
    print("Contraseña generada:", contraseña)
    
    if copiar:
        pyperclip.copy(contraseña)
        print("(Copiada al portapapeles)")
    
    return contraseña

crear_contraseña(12, True, True, True, True, copiar=True)
```
