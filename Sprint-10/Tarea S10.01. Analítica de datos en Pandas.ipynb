```python
# Nivel 1 - Punto 3
# 3) Junta las columnas Hijos y No Hijos en una sola columna, utilizando el método .apply() 
#    y definiendo una función que resuelva el problema. 
#    La columna nueva debe llamarse "Hijos" y tomar los valores "Sí" o "No"

# 3.1) Comprobamos que la cantidad de NaN en 'Fills' sea coherente con la cantidad de datos validos en 'No Fills' y viceversa

print(df_sorted["Fills"].value_counts(dropna=False))
print(df_sorted["No Fills"].value_counts(dropna=False))
```

```python
# 3.2) Creamos una funcion para normalizar los datos en una sola columna

def té_fills(row):
    if row["Fills"] == 1.0:
        return "Sí"
    elif row["No Fills"] == 1.0:
        return "No"
    else:
        return np.nan
```

```python
# 3.3) Aplicamos la función a la columna 'Fills' 

df_sorted["Fills"] = df_sorted.apply(té_fills, axis=1)

df_sorted
```

```python
# Eliminamos la columna 'No Fills'

df_sorted = df_sorted.drop(columns=["No Fills"])

#Verificamos

df_sorted
```

```python
# Nivel 1 - Punto 4
# 4) Crea una tabla resumen que permita ver el sueldo medio, mediano, mínimo y máximo por Género.

# 4.1) Verificar que tipo de dato tiene la columna 'Salari mensual'

df_sorted["Salari mensual"].dtype
```

```python
# 4.2) Cambiar el nombre de la columna 'Salari mensual' por 'Salari mensual (€)'

df_sorted = df_sorted.rename(columns={
    "Salari mensual": "Salari mensual (€)"
})

df_sorted.head(0)

```

```python
# 4.3) Eliminamos el signo (€) y el (.) de la columna 'Salari mensual (€)

df_sorted["Salari mensual (€)"] = (
    df_sorted["Salari mensual (€)"]
    .str.replace("€", "")
    .str.replace(".", "")
    .str.strip()
)


df_sorted.head(5)
```

```python
# 4.4) Convertir la columna a tipo numérico

df_sorted["Salari mensual (€)"] = pd.to_numeric(
    df_sorted["Salari mensual (€)"]
)


df_sorted["Salari mensual (€)"].dtype
```

```python
# 4.5) Realizamos los calculos solicitados

taula_salaris = (
    df_sorted
    .groupby("Gènere")["Salari mensual (€)"]
    .agg(
        sou_mig="mean",
        sou_mediana="median",
        sou_minim="min",
        sou_maxim="max"
    )
).round(2)

taula_salaris
```

```python
# 4.6) Ordena la tabla en función del sueldo medio.

taula_salaris = taula_salaris.sort_values(by="sou_mig")         #Por qué no me sale los valores NaN

taula_salaris
```

```python
# Nivel 1 - Punto 5
# 5.1) Crea una tabla resumen con el salario medio por género (filas) y país de origen (columnas). 
#      Añade las medias a los márgenes de la tabla.

taula_genere_pais = pd.pivot_table(
    df_sorted,
    values="Salari mensual (€)",
    index="Gènere",
    columns="País d'origen",
    aggfunc="mean",
    margins=True,
    margins_name="Media total"
)

taula_genere_pais = taula_genere_pais.round(2)

taula_genere_pais

```

```python

# 5.3) (EXTRA): Aplica formato condicional en la tabla para ver en un color más intenso los valores más elevados

taula_genere_pais.style.format("{:,.2f} €").background_gradient(cmap="Greens")
```

```python
# Nivel 1 - Punto 6
# 6) Crea una nueva columna que sea la fecha de nacimiento en formato Datetime a partir de las columnas día, mes y año. 

# 6.1) Cambiamos los nombres de columna porque Pandas espera que los nombres exactos sean 'year', 'month' y 'day' 
     # cuando pasamos un DataFrame para pd.to_datetime()

df_sorted.rename(
    columns={"Any":"year","Mes":"month","Dia":"day"},
    inplace=True
)


df_sorted.head()
```

```python
# 6.2) Creamos la columna fecha de nacimiento

df_sorted["Data_naixement"] = pd.to_datetime(
df_sorted[["year", "month", "day"]]
)

# Comprobamos

df_sorted[["day", "month", "year", "Data_naixement"]].head()
```

```python
# 6.3) Utilizando esta columna crea una función que dada una fecha, te calcule la edad actual a día de hoy.
from datetime import date

def calcular_edad(data_naixement):

    hoy = date.today()
    edad = hoy.year - data_naixement.year
    
    if (hoy.month, hoy.day) < (data_naixement.month, data_naixement.day):
        edad -= 1
        
    return edad
```

```python
# 6.4) Utiliza la función que acabas de crear para generar una columna nueva en DataFrame con la edad actual.

df_sorted["Edat actual"] = df_sorted["Data_naixement"].apply(calcular_edad)

df_sorted[["Nom complet", "Data_naixement", "Edat actual"]].head()
```
