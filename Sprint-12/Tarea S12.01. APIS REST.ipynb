3. Haz una petición GET sencilla:

* Muestra el código de estado de la respuesta.
* Imprime de forma clara algunos de los campos de la respuesta JSON.

```python
url_base2 = "https://restcountries.com/v3.1/all?fields=name,flags`"
```

---------------------------

```python
params = {
    "fields": "cca3,name,currencies,languages,capital,region"
}

response = requests.get(url_base2, params=params)

print("Status code:", response.status_code)
```

```python
world_data = response.json()

world_data
```

-------------

4. Convierte los datos a un DataFrame de pandas:

* Muestra las primeras filas de DataFrame.

* Opción A

```python
df_opcionA = pd.json_normalize(world_data)

df_opcionA.head(1)
```

* Opción B

```python
world_data_clean = []

for country in world_data:
    capital_list = country.get("capital")

    if capital_list and len(capital_list) > 0:
        capital = capital_list[0]
    else:
        capital = "N/A"

    world_data_clean.append({
        "code": country.get("cca3"),
        "full_name": country["name"].get("official"),
        "capital": capital,
        "region": country.get("region"),
        "currency": ", ".join(
            [v["name"] for v in country.get("currencies", {}).values()]
        ),
        "language": ", ".join(
            country.get("languages", {}).values()
        )
    })
```

```python
df_opcionB = pd.DataFrame(world_data_clean)
df_opcionB.head(1)
```

1. Utiliza la API de Open Data BCN para:

* Buscar un dataset de tu interés mediante package_search y package_show. .



```python
base_url_bcn = "https://opendata-ajuntament.barcelona.cat/data/api/action"
```

```python
search_params = {
    "q": "Bicing"
}

search_response = requests.get(f"{base_url_bcn}/package_search", params=search_params)
search_data = search_response.json()

print("Status code package_search:", search_response.status_code)
print("Número de datasets encontrados:", search_data["result"]["count"])
```
